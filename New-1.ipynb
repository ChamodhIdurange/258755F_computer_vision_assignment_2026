```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a 200x200 grid of zeros (black image)
# We use 'uint8' because standard images use 8-bit integers (0-255)
image = np.zeros((200, 200), dtype=np.uint8)

# 2. Use NumPy slicing to create a white square in the middle
# This sets rows 50-150 and columns 50-150 to 255 (white)
image[50:150, 50:150] = 255

# 3. Create a horizontal gradient using a NumPy linear space
gradient = np.linspace(0, 255, 200, dtype=np.uint8)
image[160:180, :] = gradient 

# 4. Display the result
plt.figure(figsize=(6, 6))
plt.imshow(image, cmap='gray')
plt.title("NumPy Generated Image")
plt.axis('off') # Hide pixel coordinates
plt.show()
```
